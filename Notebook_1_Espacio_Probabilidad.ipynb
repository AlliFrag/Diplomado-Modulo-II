##Probabilidad Laplaciana

Para un experimento aleatorio, se tiene que $\Omega$ es el espacio muestral (conjunto de posibles resultados).

Sea $A$ un evento, entonces definimos la probabilidad de $A$ como sigue:

$$ \mathbb{P}(A) = \frac{|A|}{|\Omega|}. $$

**Ejemplo.** Sea el experimento de lanzar una moneda 3 veces consecutivas, de manera independiente, es decir, el $i$-ésimo lanzamiento no afecta al $i+1$ lanzamiento. Es espacio muestral es:

$$ \Omega_{\text{moneda}} = \{ a,s \}$$

$$ \Omega = \Omega_{\text{moneda}} \times \Omega_{\text{moneda}} \times \Omega_{\text{moneda}} = \{(s,s,s),(s,a,s),(s,s,a),(s,a,a),(a,s,a),(a,s,s),(a,a,s),(a,a,a)\} $$
y se tiene que $|\Omega|= 2^3 = 8$.

Sean los eventos:

- $A$: todos los resultados son águila
- $S$: todos los resultados son sol
- $B$: por lo menos un resultado es sol (1 o 2 o 3 soles)

Vamos a calcular las probabilidades de $A,S$ y $E$. Para esto, sea

- $A_i$ : cae águila en el $i$-ésimo lanzamiento.

Para el evento $A$, se tiene que $A=A_1 \cap A_2 \cap A_3$


### 1. Primero vamos a crear el espacio muestral

Utilizamos el siguiente modulo (https://docs.python.org/3/library/itertools.html):

```python
import itertools
from itertools import product
```

```python
# Definimos el espacio muestral
Omega = set(product({'A','S'}, repeat=3)) # product genera tuplas de longitud 3 y set lo convierte en conjunto
Omega
```

```python
#Omega1 = set(product({'A','S'}, repeat=10))
#Omega1
```

```python
# La cardinalidad de A es:
len(Omega)
```

### 2. Ahora vamos a crear el evento A

```python
# Para el evento A_1: el primer elemento es águila
A_1 = { om for om in Omega if om[0]=='A' }
A_1
```

```python
# Para el evento A_2: el segundo elemento es águila
A_2 = { om for om in Omega if om[1]=='A' }
A_2
```

```python
# Para el evento A_3: el tercer elemento es águila
A_3 = { om for om in Omega if om[2]=='A' }
A_3
```

```python
# Se tiene que A es:
A = A_1.intersection(A_2.intersection(A_3))
A
```

### 3. Ahora calculamos la probabilidad de $A$

```python
card_A = len(A)
```

```python
P_A = card_A / len(Omega)
P_A
```

Si quiero expresar mi resultado em una fracción, utilizo https://docs.python.org/3/library/fractions.html para números racionales

```python
from fractions import Fraction
```

```python
# Definimos la función de probabilidad Laplaciana
def p(E,Omega):
    p = Fraction(len(E),len(Omega))
    return p
```

```python
Pro_A = p(A,Omega)
Pro_A
```

## Probabilidad Condicional

Definimos la probabilidad condicional de que ocurra un evento $A$ **sabiendo** que ocurre un evento $B$ de la siguiente manera:

$$ \mathbb{P}(A \mid B) = \frac{\mathbb{P}(A\cap B)}{\mathbb{P}(B)} = \frac{|A\cap B|}{|B|},$$
donde $\mathbb{P}(B)>0$.

```python
# Definimos la función de probabilidad condicional
def prob_cond(E,F):
    return len(E & F) / len(F)
```

Queremos calcular la probabilidad condicional de los eventos:

- $B$: el primer lanzamiento es águila
- $C$: dos de los resultados son águila

Queremos calcular $\mathbb{P}(B\mid C)$.

```python
# Para el evento B:
B = {om for om in Omega if om[0]=='A'}
B
```

```python
# Para el evento C:
C = {om for om in Omega if om.count('A')==2 }
C
```

```python
prob_cond(B,C)
```

```python
prob_cond(C,B)
```

## Independencia de eventos

Decimos que dos eventos son independientes si
$$ \mathbb{P}(A \cap B) = \mathbb{P}(A)\mathbb{P}(B)$$

```python
def prob(E):
    return len(E) / len(Omega)
```

```python
# Definimos la función de probabilidad condicional
def indep(E,F):
    return prob(E & F) == prob(E)*prob(F)
```

```python
indep(B,C)
```


# Información adicional

## Espacio de probabilidad

Un espacio de probabilidad se representa mediante la terna:

$$
(\Omega,\mathcal{F},P)
$$

- $\Omega$: espacio muestral.
- $\mathcal{F}$: colección de eventos.
- $P$: función que asigna probabilidades.

## Axiomas de Kolmogorov

1. $P(A)\geq 0$.
2. $P(\Omega)=1$.
3. Si $A_1,A_2,\ldots$ son eventos mutuamente excluyentes:

$$
P\left(\bigcup_i A_i\right)=\sum_i P(A_i)
$$



## Probabilidad condicional e independencia

La probabilidad condicional se define como:

$$
P(A\mid B)=\frac{P(A\cap B)}{P(B)},\qquad P(B)>0
$$

Dos eventos son independientes cuando:

$$
P(A\cap B)=P(A)P(B)
$$

En ese caso, conocer que ocurrió uno no cambia la probabilidad del otro.



## Enfoques de probabilidad

| **Enfoque** | **¿En qué se basa?** | **Ejemplo** |
|-------------|----------------------|-------------|
| **Clásico** | Casos favorables / casos posibles. | Lanzar un dado. |
| **Frecuentista** | Repetir un experimento muchas veces. | Lanzar una moneda 1000 veces. |
| **Geométrico** | Comparar áreas, longitudes o volúmenes. | Lanzar un dardo a un tablero. |

**Diferencia:**  
- **Clásico:** cuenta casos equiprobables.  
- **Frecuentista:** usa la frecuencia de los resultados.  
- **Geométrico:** usa proporciones de medidas (área, longitud o volumen).



## Problema del encuentro

Dos personas llegan al azar entre las 12:00 y la 1:00 y cada una espera 15 minutos. Si $X$ y $Y$ son sus tiempos de llegada, se encuentran cuando:

$$
|X-Y|\leq 15
$$

Por el enfoque geométrico:

$$
P(\text{encuentro})=\frac{7}{16}=0.4375
$$


```python
# Simulación del problema del encuentro
import numpy as np

np.random.seed(123)
n = 100000
x = np.random.uniform(0, 60, n)
y = np.random.uniform(0, 60, n)

prob_simulada = np.mean(np.abs(x-y) <= 15)
print("Probabilidad teórica:", 7/16)
print("Probabilidad simulada:", prob_simulada)
```


### Interpretación

El resultado simulado debe aproximarse a $0.4375$. La pequeña diferencia se debe al carácter aleatorio de la simulación.



## Ejemplo: extracción de una carta

Este es un ejemplo del enfoque clásico porque todas las cartas tienen la misma posibilidad de ser seleccionadas.
